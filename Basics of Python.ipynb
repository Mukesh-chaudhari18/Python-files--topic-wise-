# Data Types in Python 

### 1 . Numbers - 
- **Int** - Whole Value , - infinity to + Infinity 
- **Float** - Decimal Point value , -infinity.00 - +infinity.00
- **Complex** - Number + character - eg. 5j+6  , (not used for mathematical operations)

```python
# Direct assigning
num1 = -34
num2 = -45.3
num3 = 4j+6        # e,j
```

```python
# Using constrctor method 
a = int(45.4444444)
b = float(32)

print(a)
print(b)
```

```python
print(type(num1))
print(type(num2))
print(type(num3))
```

```python
n1 = 20
n2 = 6
```

```python
# Addition - "+"
n1 + n2
```

```python
# Substraction - "-"
n1 - n2
```

```python
# Multiplication - "*"

n1 * n2
```

```python
# Division - "/"

n1 / n2
```

```python
# Floor Division - "//"
n1 // n2
```

```python
# Exponential Values - "**"  - X power n  - first value will be x and ** after this power value (n) 

n1 ** n2
```
