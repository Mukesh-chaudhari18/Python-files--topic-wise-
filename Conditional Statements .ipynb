### Conditional Statements 
- Allows a Programmer to write logical Programmes that checks the condition and Execute the block of code if the condition is true and ignores the code if the condition is False 
- if - First condition  
- elif - Further conditions and it can be used as many as possible 
- else  - When everything is False then else block will be executed 

```python
if True:
    print("if block executed")
```

```python
if False:
    print("if block ignored")
```

```python
if True:
    print("If block got exectued")
elif True:
    print("Elif block got executed")
else:
    print("Else block got exectured")
```

### Q5. Check if the year is leap year or not 

```python
year = int(input("Enter A year"))
if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not A leap Year")   
```

### Q6. BMI calculator 

```python
height = float(input("Enter your height"))
weight = float(input("Enter Your Weight"))
BMI = weight / (height **2)
print(BMI)
if BMI < 18.5 :
    print("Underweight")
elif BMI < 25:
    print("Normal Weight")
elif BMI < 30:
    print("Over Weight")
else:
    print("Obese")
```

### Q8. Electric Bill Calculator - 
- for first 100 unit - Rs 5 / unit 
- for next 100 unit - Rs 7 / unit 
- Above 200 unit - Rs 10 / unit
- above 300 units - Rs 15 / unit
- above 400 units - RS 19 / unit 

```python
units = float(input("Enter Units Used"))
if units < 100 :
    bill = units * 5
elif units < 200 :
    bill = ( 100 * 5) + ( units - 100 ) * 7
elif units < 300: 
    bill = ( 100 * 5) + ( 100 * 7 ) + ( units - 200) * 10
elif units < 400 :
    bill = (100 * 5) + (100*7) + (100*10) + (units -300) * 15
else:
    bill = (100*5) + (100*7) + (100*10) + (100*15) + (units - 400 ) * 19


print(f"The Amount of Total Electric Bill for { units }  is { bill } ")
```
