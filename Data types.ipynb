#### Indexing - 
- Indexing is numbering the values in sequence 
- Postive Indexing [ 0 - n-1] and negative indexing [ -n - -1]

- Syntax for indeing - Var[Index_number] 

```python
a = "Never Give Up"
len(a)
```

#### Slicing / index range 
- Method of Returning Some part of data / text using index range 
- Syntax - variable[start_index : Stop_index ]  or variable[Start_index: stop_index : step ] (skip values)

```python
data = "Python is Fun"
data
```

```python
data[:]   # return all data  [ : ] - considers start with all and end with all
```

```python
data[3:]  # start with 3rd index and include all  [ 3: ] 
```

```python
data[1:4]  # start with 1st index and stop on 4th index 
```

```python
data
```
