### Two types of Loops -
- While Loop - Infinity loop
- For Loop - Iterative loop 

```python
while False :
    print("Infinity")
```

```python
x = 15
while x < 19 :
    print(x)
    x += 1 
```

```python
while True:
    num = int(input("Enter Zero to stop the loop"))
    if num == 0 :
        print("Loop Ended")
        break 
```

### For Loop - Iterative loop 

```python
text = input("enter something")
for i in text:
    print(i.upper())  
```

```python
for i in range(1,10):
    if i % 2 == 0 : 
        continue 
    print(i)
```

#### Q1. Print the even values from 0-100 

```python
l = []
for i in range(0,100):
    if i % 2 == 0:
        l.append(i)
print(l)   
```

#### Q2. Print the odd values from 0- 100

```python
l = [] 
for i in range(0,100):
    if i %2 != 0:
        l.append(i)
print(l)
```

#### Q3. Take a Number from user and return the sum of all numbers 

```python
num = int(input("enter a number"))
total = 0
for i in range(1,num+1):
    total += i 
print(total)
```

#### Q4. Take a Number from user and Return it's table 

```python
num = int(input("Enter A number"))
for i in range(1,11):
    print(i*num)
```

#### Q5. Take a List and return the smallest and largest elements from list 

```python
l1 = [ -34,324,45,23,0,345,2345,2492,552,242,1093,242,234,46543,-132]
large = l1[0]
small = l1[0]
for i in l1:
    if i > large :
        large = i
    elif i < small :
        small = i 
print(large)
print(small)
```

#### Q6. Take a String from user and reverse the string without using [::-1]

```python
text = input("Enter a Text")
reverse = ''
for i in text:
    reverse = i + reverse 
print(reverse)
```

#### Q7. Take a Number from user and return it's Factorial Number 

```python
num = int(input("Enter A number to return it's factorial"))
fact = 1
for i in range(1, num+1):
    fact *= i
print(f"The Factorial for {i} is {fact}")
    
```

#### Q8. Count the vowels in the user string 

```python
text = input("Enter A text")
count = 0 
for i in text.lower():
    if i in "aeiou":
        count += 1
print(f" The Total count of vowels is { count }")
```

#### Q8. Pattern Printing 

```python
num = int(input("Enter the number"))
for i in range(1 , num+1):
    print( "*" * i)
```

```python
num = int(input("Enter the number"))
for i in range(num+1,0,-1):
    print( "*" * i)
```

```python
num = int(input("Enter the number"))
for i in range(1 , num+1):
    print( "*" * i)
for i in range(num,0,-1):
    print( "*" * i)
```
