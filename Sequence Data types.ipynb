#### List - 
- In built Data type
- denoted by list() or [  ] 
- Ordered data type - allows indexing and slicing
- Duplicated can be stored
- Mutable data Type - elements can be added , removed or replaced.
- Data Types allowed to stored - All data types 

```python
# using [] direct assignation
number = [ 1,2,3,4,5,6]
text = [ 'A' , 'B' , 'C', 'D', 'E', 'F' ]
mixed = [ True, 23 , "python", 45.67  , [1,2,3] , (45,3,2) , {1,2,3} , {"Name" : "Human"},None , bin(232)] 

print(number , type(number))
print(text , type(text))
print(mixed , type(mixed))
```

```python
# using constructor functions - list()
a = list("Make a list")   # string
b = list((1,2,"python", "create" , "a" ,"list"))  #tuple
c = list(range(0,50))   #range
d = list([1,2,3,4,5,6])  #list
e = list({1,2,3,1,1,1,4,5,5})  #set

print(a)
print(b)
print(c)
print(d)
print(e)
```

#### Functions to Add Elements in the list 
- append(item) - Add the item in the end 
- extend(iterable) - Merges the iterable into the existing list 
- insert(i,x) - at Certain index value it adds the X value (but does not remove the existing value)

```python
a = [1,2,3]
```

```python
a.append(5)    # in the end as it is it adds the item 
a.append("Add")
a.append([45,45,45])
a.append((1,2,3))
a.append({4,23,543,42})
```

# Tuple 
- In built data types
- Denoted by () or tuple()
- Ordered - indexing and slicing is allowed
- Duplicates can be stored
- Immutable in nature - cannot add items , change item or remove something
- Data types allowed - all data types 

```python
tup = 1,2,3,4,5,5,4,3,3
tup1 = ( 1,3,32,33,45)
tup2 = tuple(["apple" , "kiwi" , "orange",  "mango","grapes","watermalon" , "banana" ,"dragonfruit" ])
tup3 = (34,)
print(tup, type(tup))
print(tup1, type(tup1))
print(tup2, type(tup2))
print(tup3, type(tup3))
```
