### **➤ Important String Methods**

Method	           Purpose	                     Example
lower()	      Converts to lowercase	           text.lower()
upper()	      Converts to uppercase	           text.upper()
title()	      Capitalizes each word	           text.title()
strip()	       Removes extra spaces	           text.strip()
replace()	    Replace characters	           text.replace('a', 'b')
split()	      Split into list of words	       text.split()

```python

text = "   Python is FUN!   "
print(text.strip())   # Python is FUN!
print(text.replace("FUN", "AWESOME"))
```
