## 📘 What is a Tuple?

- A **tuple** is an ordered collection of elements.
- It is **immutable**, meaning **you cannot modify** (add, remove, change) items after creation.
- Defined using **round brackets** `()`.

> Use tuples when you want to **protect data from accidental modification**.


```python
tup1 =  1,2,3,2,3,3,2,4,3,2,3
```

```python
tup1
```

```python
print(type(tup1))
```

```python
tup2 = tuple(["Tuple" , 12.4 , 34, True , [1,2,3]])
```

```python
tup2
```

```python
# Example of tuple
my_tuple = (10, 20, 30)
print(my_tuple)       # (10, 20, 30)
```

## 🛠️ Creating Tuples

You can create tuples with or without parentheses, with different types of data.


```python
tup = 1,2,3
```

```python
print(tup)
```

```python
# Creating tuples
t1 = (1, 2, 3)
t2 = 'a', 'b', 'c'         # without parentheses
t3 = ()                    # empty tuple
t4 = (5,)                  # single-element tuple (note the comma!)

print(t1)
print(t2)
print(t3)
print(t4)
print(type(t2))
```

## 🧰 Tuple Built-in Functions

| Function | Description |
|----------|-------------|
| `len()` | Length of the tuple |
| `max()` | Max value (numeric only) |
| `min()` | Min value |
| `sum()` | Sum of elements (if numeric) |
| `type()` | Data type |


```python
nums = (5, 10, 15)

print(len(nums))        # 3
print(max(nums))        # 15
print(min(nums))        # 5
print(sum(nums))        # 30
print(type(nums))       # <class 'tuple'>
```

## 📌 Tuple Methods

| Method | Description |
|--------|-------------|
| `count(x)` | Counts the number of times x appears |
| `index(x)` | Returns the first index of value x |


```python
tup1 = (1,1,1,2,3,5,3,2,3,4,4,"hello","tuple", "is fun to learn","tuple")
```

```python
tup1.count("tuple")
```

```python
tup1.index(1,2)
```

## 🧠 Tuple vs List

| Feature        | Tuple           | List             |
|----------------|------------------|------------------|
| Syntax         | `(1, 2, 3)`       | `[1, 2, 3]`       |
| Mutable        | ❌ No             | ✅ Yes            |
| Methods        | Few              | Many             |
| Use Case       | Fixed Data       | Changing Data    |
| Performance    | Faster           | Slower           |
