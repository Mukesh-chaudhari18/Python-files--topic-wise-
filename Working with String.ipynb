```python
# ✅ replace() with multiple replacements

text = "hello world"
#print(text.replace("l", "*"))      # he**o wor*d
print(text.replace("hello", "Python"))  # hello Python
```

```python
# ✅ join() and split()

#words = ["Data", "Science","machinelearning"]
#print(" ".join(words))         # Data Science

sentence = "AI is the future"
print(sentence.split(" "))       # ['AI', 'is', 'the', 'future']
```

```python
# ✅ startswith() and endswith()

url = "https://openai.com"
print(url.startswith(".com"))    # True
print(url.endswith("https"))       # True
```

```python
# ✅ partition()

email = "user@gmail.com"
print(email.partition("."))   # ('user', '@', 'gmail.com')
```
