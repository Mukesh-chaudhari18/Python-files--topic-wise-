```python
# 1.union or | - it returns all unique elements from both sets 
a = { 1,2,3,4,5} 
b = { 4,5,6,7,8}
c = { 3,56,34,23,543} 
print(a.union(b).union(c))
print( a | b | c)
```

```python
# 2. Intersection or & - it returns All matching elements from both sets , & 
a = { 1,2,3,4,5} 
b = { 4,5,6,7,8}
print(a.intersection(b))
print( a & b)
```

### Dictionary 
- Denoted in { key : value } or dict() 
- Ordered Data Type - Key is used for Orderness
- Mutable Data type - Add , modify and remove the elements 
- Mapping Data Type
- Data type stored
- key - int, float , string , tuple
- value - all data types
- Duplicates
- Key is not allowed to be duplicate
- Value can be duplicates
