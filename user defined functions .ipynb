### function(*argument) 
- Allows to insert many arguments
- used generally for cumulative operations 

```python
def total(*num):
    return sum(num)
```

```python
total(4,2,1,3,4,2,2,3,4,56,2,2,2,2,2,2,2)
```

```python
def avg(*num):
    return sum(num)/len(num)
```

```python
avg(34,23,67,45)
```

```python
avg(55,53)
```

### function( ** arguments ) 
- keyword arguments
- used to store data like key and value pair
- Internally it stores the dictionary data
- When this is argued - argument rules - function( para1 = value1 , para2 = value2, .....) 

```python
def user(**info):
    print( info ) 
```

```python
user(name="Viraj" , age=23,job_role = "data scientist" )
```

```python
def user(**info):
    for i,j in info.items():
        print(f" {i} : {j}")
```

```python
user( name = "Dharmesh" , job_role = "Manager" , city = "Vadodara" , age = 45 , salary = 45600 )
```

### lambda Functions - 
- Nameless functions / Annonymous Functions
- Used for single line code or logic
- used with
  - Map - Operation on all elements of iterable
  - filter - applying filteration on any iterable
  - reduce - Cumultive operation 
- syntax - lambda arguemnts : expression 

```python
sq = lambda a : a**2 
cube = lambda x : x**3
check = lambda z : "Even" if z % 2==0 else "Odd"
```

```python
sq(45)
```

```python
cube(3)
```

```python
check(98)
```

## Lambda + map - iterates to every element and does the operation 

```python
l1 = [ -2,67,44,3,23,32 ] 
list(map(lambda x : x**2 , l1 ))
```

```python
# Square of all values from 1 - 50
print(list(map(lambda x : x**2 , range(1,51))))    
```

```python
# Cube of all values from 1-50 
print(list(map(lambda x : x**3 , range(1,51))))   
```

```python
print(list(map(lambda x : x.title() , ['daTA SCIenCe' , 'DeEP lEArNiNG' , 'GeN aI'])))   
```

### Lambda + Filter 
- Used for filtering the data 

```python
### even values from 0 -100 
print(list(filter(lambda x : x % 2 ==0 , range(1,100))))
```

```python
### Odd values from 0 -100 
print(list(filter(lambda x : x % 2 !=0 , range(1,100))))
```

```python
### text filter only data starts with A 
print(list(filter(lambda x : x.startswith("A"),['Lily','Amigo','Aman','Zebra','Apple','Binary','ML'])))
```

### Lambda + reduce 
- reduce is called through functntools 
- used for combining the output into single value 

```python
from functools import reduce 
```

```python
l1 = [ 23, 33, -11 ,23 ,5 ,3] 
results = reduce( lambda x,y : x+y , l1)
results
```

```python
l1 = [ 23, 33, -11 ,23 ,5 ,3] 
results = reduce( lambda x,y : x-y , l1)
results
```

```python
reduce(lambda x , y : x*y , range(1,10))
```
